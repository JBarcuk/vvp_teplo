```python
import json
import numpy as np
import simulation
```

```python
# load data/cihly/parametry.json
with open('data/cihly/parametry.json') as f:
    data = json.load(f)
# print data
print(data)
dx = data['dx']
dy = data['dy']
dt = data['dt']
print(f"dx = {dx}, dy = {dy}, dt = {dt}")

lam = np.load('data/cihly/LAM.npy')
rho_c = np.load('data/cihly/rho_c.npy')
u = np.load('data/cihly/U_initial.npy')

amountOfSeconds = 3600
amountOfCycles = int(amountOfSeconds / dt)

u_cycle = simulation.simulateByCycle(u, lam, rho_c, dx, dy, dt, amountOfCycles)
u_matrix = simulation.simulateByMatrix(u, lam, rho_c, dx, dy, dt, amountOfCycles)

simulation.plot(u_cycle, f"Simulace tepla po {amountOfSeconds}s za pomocí cyklu")
simulation.plot(u_matrix, f"Simulace tepla po {amountOfSeconds}s za pomocí matice")
```
